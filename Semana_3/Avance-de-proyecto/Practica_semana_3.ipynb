## 1. Tema 1: Algoritmo y modelo Entrada-Proceso-Salida (EPS)

Para empezar, identifiqué qué información necesito ingresar, qué operaciones debo realizar y qué resultados quiero obtener en mi programa.

En mi proyecto lo organicé de esta manera:

- **Entrada:** ingreso las cantidades de conchas, bolillos, donas, orejas y rebanadas de pastel que se vendieron.
- **Proceso:** multiplico la cantidad de cada producto por su precio y después sumo todos los importes.
- **Salida:** muestro cuánto cuesta cada producto según la cantidad vendida y cuánto debe pagar el cliente en total.

Para entender mejor este proceso, voy a utilizar las mismas cantidades que aparecen en la captura de mi programa.

```python
precios = {
    "Concha": 15.00,
    "Bolillo": 4.00,
    "Dona": 18.00,
    "Oreja": 16.00,
    "Rebanada de pastel": 45.00
}

cantidades = {
    "Concha": 98,
    "Bolillo": 100,
    "Dona": 25,
    "Oreja": 35,
    "Rebanada de pastel": 12
}

total_compra = 0.0

for producto, cantidad in cantidades.items():
    importe = precios[producto] * cantidad
    total_compra += importe

    print(
        f"{producto}: {cantidad} x "
        f"${precios[producto]:.2f} = ${importe:.2f}"
    )

print(f"Total a pagar: ${total_compra:.2f}")
```

**Resultado**

Al ejecutar este código, obtengo los siguientes importes:

- Conchas: $1,470.00.
- Bolillos: $400.00.
- Donas: $450.00.
- Orejas: $560.00.
- Rebanadas de pastel: $540.00.

Al sumar todos los importes, obtengo un total de **$3,420.00**.

Con este ejercicio entiendo mejor cómo puedo ingresar datos, realizar operaciones y mostrar el resultado de una venta utilizando Python.

## 3. Tema 3: Variables, tipos de datos y operadores

En mi programa utilizo variables para guardar los datos que necesito al registrar una venta.

Los tipos de datos que utilizo son:

- **`str`:** lo utilizo para guardar texto, como el nombre de un producto.
- **`int`:** lo utilizo para guardar números enteros, como la cantidad de piezas vendidas.
- **`float`:** lo utilizo para guardar precios e importes que pueden tener decimales.
- **`bool`:** lo utilizo para representar si una condición es verdadera o falsa, por ejemplo, para saber si una venta es válida.

También utilizo operadores para hacer las cuentas y comprobar diferentes condiciones. Esto me ayuda a calcular los importes y a revisar que los datos que ingreso sean correctos.

```python
# Variables y tipos de datos

nombre_producto = "Concha"       # str
cantidad = 3                     # int
precio = 15.00                   # float
venta_valida = cantidad > 0      # bool

importe = cantidad * precio

print("Producto:", nombre_producto)
print("Cantidad:", cantidad)
print("Precio:", precio)
print("¿La venta es válida?:", venta_valida)
print("Importe:", importe)

# Operadores matemáticos
print("Suma:", 10 + 5)
print("Resta:", 10 - 5)
print("Multiplicación:", 10 * 5)
print("División:", 10 / 5)

# Operadores relacionales y lógicos
print("¿La cantidad es igual a cero?:", cantidad == 0)

print(
    "¿La cantidad y el precio son positivos?:",
    cantidad > 0 and precio > 0
)
```

**Explicación**

En este ejercicio utilicé diferentes tipos de datos para guardar la información de un producto. También hice operaciones matemáticas y comparaciones.

Por ejemplo, multipliqué la cantidad por el precio para obtener el importe. Después, comprobé si la cantidad y el precio eran mayores que cero.

Así puedo entender mejor cómo utilizar las variables y los operadores en mi proyecto.

## 4. Tema 4: Entradas y salidas

En mi programa necesito pedirle información al usuario y después mostrarle los resultados.

Para hacerlo, utilizo `input()`, que me permite pedir datos, y `print()`, que me permite mostrar mensajes y resultados en la pantalla.

Por ejemplo, puedo pedir el nombre de un producto y la cantidad que se vendió. Después, puedo revisar si la cantidad es válida.

También debo tomar en cuenta que `input()` recibe los datos como texto, por lo que necesito convertir la cantidad a un número entero antes de utilizarla en las operaciones.

```python
producto = input("Escribe el nombre de un producto: ")
cantidad_texto = input("Escribe la cantidad vendida: ")

try:
    cantidad = int(cantidad_texto)

    if cantidad >= 0:
        print(f"Se registraron {cantidad} piezas de {producto}.")
    else:
        print("La cantidad no puede ser negativa.")

except ValueError:
    print("Debes escribir una cantidad con número entero.")
```

**Explicación**

En este ejercicio le pedí al usuario el nombre de un producto y la cantidad vendida. Después, convertí la cantidad a un número entero para poder revisarla.

Si la cantidad es negativa, el programa muestra un mensaje para indicar que no es válida. Si escribo letras en lugar de un número entero, también aparece un mensaje que me pide corregir el dato.

Con esto entiendo cómo puedo pedir información y revisar los datos antes de utilizarlos.

## 5. Tema 5: Estructuras de decisión (`if`, `elif` y `else`)

En mi programa necesito tomar decisiones dependiendo de los datos que ingreso. Para eso utilizo las estructuras `if`, `elif` y `else`.

- Utilizo `if` para comprobar una condición.
- Utilizo `elif` cuando quiero revisar otra condición.
- Utilizo `else` cuando ninguna de las condiciones anteriores se cumple.

Estas estructuras me ayudan a decidir qué mensaje mostrar o qué acción realizar según la cantidad que se haya ingresado.

```python
cantidad_prueba = 2

if cantidad_prueba < 0:
    print("La cantidad no es válida.")

elif cantidad_prueba == 0:
    print("No se vendieron piezas de este producto.")

else:
    print("La cantidad se puede utilizar para calcular el importe.")
```

**Explicación**

En este ejercicio utilicé condiciones para revisar la cantidad que tengo guardada en la variable.

Si la cantidad es negativa, el programa me indica que no es válida. Si la cantidad es cero, me dice que no se vendieron piezas. Si la cantidad es positiva, me indica que puedo utilizarla para calcular el importe.

Así puedo hacer que mi programa responda de diferente manera según los datos que recibe.

## 6. Tema 6: Ciclo `while`

El ciclo `while` me permite repetir instrucciones mientras se cumpla una condición.

En mi proyecto lo utilizo para mantener el menú activo y permitir que se registren varias ventas sin tener que cerrar el programa después de cada una.

También puedo utilizarlo para revisar una cantidad y repetir la validación cuando el usuario introduce un valor negativo.

```python
cantidad = -1

while cantidad < 0:
    print("La cantidad no puede ser negativa.")

    # Asigno una cantidad válida para terminar el ejemplo.
    cantidad = 2

print("Cantidad aceptada:", cantidad)
```

**Explicación**

En este ejemplo comencé con una cantidad negativa. Como esa cantidad no es válida, el ciclo muestra un mensaje.

Después, cambié la cantidad a un número positivo y el ciclo terminó. Finalmente, el programa mostró la cantidad aceptada.

En mi programa de ventas puedo utilizar esta misma idea para volver a pedir una cantidad cuando sea necesario corregirla.

Para hacerlo, puedo utilizar `pdb`, una herramienta de Python que me permite detener el programa y revisar los valores de las variables.

En mi proyecto puedo utilizarla para comprobar si el precio y la cantidad son correctos antes de calcular el importe de un producto.

## 8. Tema 8: Ciclo `for`

El ciclo `for` me permite recorrer varios elementos sin tener que escribir las mismas instrucciones para cada uno.

En mi proyecto lo utilizo para recorrer los productos que se vendieron, multiplicar sus cantidades por sus precios y sumar los importes.

Esto me ayuda a organizar mejor el código, porque puedo realizar la misma operación con todos los productos mediante un solo ciclo.

```python
precios = {
    "Concha": 15.00,
    "Bolillo": 4.00,
    "Dona": 18.00,
    "Oreja": 16.00,
    "Rebanada de pastel": 45.00
}

cantidades = {
    "Concha": 98,
    "Bolillo": 100,
    "Dona": 25,
    "Oreja": 35,
    "Rebanada de pastel": 12
}

total_dia = 0.0

for producto, cantidad in cantidades.items():
    importe = precios[producto] * cantidad
    total_dia += importe

    print(f"{producto}: ${importe:.2f}")

print(f"Total acumulado de la venta de ejemplo: ${total_dia:.2f}")
```

**Explicación**

En este ejercicio recorrí los productos utilizando un ciclo `for`. Para cada uno, multipliqué la cantidad vendida por su precio y sumé el importe al total.

Al terminar el ciclo, obtuve un total de **$3,420.00**.

Con este ejercicio entiendo cómo puedo realizar cálculos para varios productos sin tener que repetir manualmente las mismas instrucciones.

## 9. Prueba integrada: registro y cálculo de una venta

En esta prueba voy a juntar varios de los temas que repasé anteriormente. Voy a utilizar variables para guardar los precios y las cantidades, operadores matemáticos para calcular los importes, un ciclo `for` para recorrer los productos y una condición para comprobar si se vendió al menos un producto.

Para que la prueba esté relacionada con mi proyecto, voy a utilizar las mismas cantidades que aparecen en la captura de mi programa de registro de ventas.

```python
precios = {
    "Concha": 15.00,
    "Bolillo": 4.00,
    "Dona": 18.00,
    "Oreja": 16.00,
    "Rebanada de pastel": 45.00
}

cantidades = {
    "Concha": 98,
    "Bolillo": 100,
    "Dona": 25,
    "Oreja": 35,
    "Rebanada de pastel": 12
}

total_compra = 0.0
hay_productos = False

for producto, cantidad in cantidades.items():
    if cantidad > 0:
        hay_productos = True
        importe = precios[producto] * cantidad
        total_compra += importe

        print(
            f"{producto}: {cantidad} x "
            f"${precios[producto]:.2f} = ${importe:.2f}"
        )

if hay_productos:
    print(f"Venta válida. Total a pagar: ${total_compra:.2f}")
else:
    print("No se registró la venta porque no se eligieron productos.")
```

**Resultado de la prueba**

Al ejecutar el código con las cantidades de mi ejemplo, obtengo los siguientes resultados:

- Conchas: $1,470.00.
- Bolillos: $400.00.
- Donas: $450.00.
- Orejas: $560.00.
- Rebanadas de pastel: $540.00.

El total de la compra es de **$3,420.00**.

Con esta prueba puedo comprobar el cálculo de los importes y relacionar las variables, los operadores, los ciclos y las condiciones con mi proyecto de registro de ventas.

```python
PRECIO_MENOR = 0.0
PRECIO_NINA = 30.0
PRECIO_GENERAL = 45.0

DESC_ESTUDIANTE = 0.10
DESC_PROFESORA = 0.10
DESC_ADULTO_MAYOR = 0.12
DESC_NINGUNO = 0.0

total_recaudado = 0.0
total_descuentos = 0.0
visitantes_atendidos = 0

while True:
    opcion_compra = input(
        "\n¿Quisiera hacer una compra? ('si' o 'fin'): "
    )


    if opcion_compra != "si":
        break

    num_boletos = int(input("¿Cuántos boletos desea?: "))
    subtotal_compra = 0.0

    for b in range(1, num_boletos + 1):


        while True:
            try:
                edad = int(input(f"  -> Edad del visitante #{b}: "))

                if 0 <= edad <= 110:
                    break
                else:
                    print(
                        "  [Error] Edad no válida (0 a 110), "
                        "intente de nuevo."
                    )
            except ValueError:
                print("  [Error] Ingrese una edad válida.")

        tipo_texto = ""
        precio_base = 0.0
        porcentaje_desc = 0.0


        if edad < 3:
            precio_base = PRECIO_MENOR
            porcentaje_desc = DESC_NINGUNO
            tipo_texto = "Menor de 3 años (Gratis)"


        elif edad < 18:
            precio_base = PRECIO_NINA
            porcentaje_desc = DESC_NINGUNO
            tipo_texto = "Niña/Adolescente (Sin descuento)"

        else:
            precio_base = PRECIO_GENERAL

            print(
                "    Tipo Visitante: "
                "1) Estudiante (10%)  "
                "2) Profesora (10%)  "
                "3) Ninguno (0%)"
            )

            tipo_op = input("    Seleccione opción (1-3): ")

            if tipo_op == "1":
                tipo_texto = "Estudiante"
            elif tipo_op == "2":
                tipo_texto = "Profesora"
            else:
                tipo_texto = "General"


            if edad >= 60:
                porcentaje_desc = DESC_ADULTO_MAYOR
                tipo_texto += " (Adulto Mayor 12%)"
            else:
                if tipo_op == "1":
                    porcentaje_desc = DESC_ESTUDIANTE
                elif tipo_op == "2":
                    porcentaje_desc = DESC_PROFESORA
                else:
                    porcentaje_desc = DESC_NINGUNO


        descuento_monto = precio_base * porcentaje_desc
        pago_boleto = precio_base - descuento_monto

        subtotal_compra += pago_boleto
        total_recaudado += pago_boleto
        total_descuentos += descuento_monto
        visitantes_atendidos += 1


        print(
            f"  [Ticket] Visitante {b} | Edad: {edad} | "
            f"{tipo_texto} | Base: ${precio_base:.2f} | "
            f"Desc: ${descuento_monto:.2f} | "
            f"A pagar: ${pago_boleto:.2f}"
        )

    print(
        f"--- Subtotal de esta compra: "
        f"${subtotal_compra:.2f} ---"
    )


print("\n" + "=" * 30)
print("=== CORTE DE CAJA FINAL ===")
print(f"Visitantes atendidos: {visitantes_atendidos}")
print(f"Total descuentos: ${total_descuentos:.2f}")
print(f"TOTAL EN CAJA: ${total_recaudado:.2f}")
print("=== FIN - TAQUILLA CERRADA ===")
```

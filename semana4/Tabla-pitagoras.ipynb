## Introducción

En esta actividad voy a desarrollar una tabla de Pitágoras del 1 al 10 utilizando una matriz, que en Python puedo representar mediante una lista que contiene otras listas.

Mi objetivo es construir la tabla utilizando ciclos `for`, mostrar todos sus valores de una manera ordenada y crear una función que me permita consultar cualquier resultado sin tener que realizar nuevamente la multiplicación.

También voy a utilizar funciones, ciclos `while`, entradas de datos y condiciones para validar que los números que ingrese el usuario estén dentro del rango permitido.

Al finalizar, voy a realizar cuatro casos de prueba para comprobar que mi programa funcione correctamente y que pueda mostrar los resultados esperados.

## PARTE A. CONSTRUCCIÓN DE LA MATRIZ

En esta primera parte voy a realizar mi tabla de Pitágoras desde cero.

Para hacerlo, voy a utilizar una matriz de 10 filas por 10 columnas. Esto significa que voy a guardar 100 resultados diferentes, comenzando desde la multiplicación de 1 por 1 y terminando con la multiplicación de 10 por 10.

Para construirla voy a utilizar dos ciclos `for`:

- El primer ciclo va a recorrer los números del 1 al 10 y me va a ayudar a crear cada fila.
- El segundo ciclo va a recorrer nuevamente los números del 1 al 10 para calcular los valores que van dentro de cada fila.
- Voy a utilizar `append()` para agregar cada resultado a su fila y después agregar la fila completa a mi matriz.

En esta parte sí voy a utilizar el operador `*`, porque necesito calcular los valores que voy a guardar dentro de la tabla.

De esta manera, mi matriz tendrá todos los resultados organizados y posteriormente podré consultarlos sin necesidad de volver a multiplicar.

```python

# PARTE A: CONSTRUCCIÓN DE LA MATRIZ DE PITÁGORAS

# Primero creo una lista vacía.
# Aquí voy a guardar todas las filas que formen mi tabla.
tabla_Pitagoras = []

# Este primer ciclo recorre los números del 1 al 10.
# Cada número representa una fila de mi tabla.
for fila in range(1, 11):

    # Creo una lista vacía para guardar los resultados
    # correspondientes a la fila que estoy construyendo.
    lista_fila = []

    # Este segundo ciclo recorre los números del 1 al 10.
    # Cada número representa una columna de mi tabla.
    for columna in range(1, 11):

        # Multiplico el número de la fila por el número
        # de la columna para obtener el resultado.
        producto = fila * columna

        # Agrego el resultado a la lista de la fila actual.
        lista_fila.append(producto)

    # Cuando termino de recorrer todas las columnas,
    # agrego la fila completa a mi matriz principal.
    tabla_Pitagoras.append(lista_fila)

# Muestro un mensaje para comprobar que la matriz
# se construyó correctamente.
print("Mi matriz de Pitágoras se construyó correctamente.")

# Muestro la cantidad de filas que contiene mi matriz.
print("Cantidad de filas:", len(tabla_Pitagoras))

# Muestro la cantidad de columnas de la primera fila.
print("Cantidad de columnas:", len(tabla_Pitagoras[0]))
```

## PARTE B. FUNCIÓN PARA MOSTRAR LA TABLA

Ahora voy a crear una función llamada `imprimir_tabla(tabla)`. Su objetivo es mostrar todos los números que guardé anteriormente en mi matriz.

Para esta función voy a utilizar nuevamente dos ciclos `for`:

- El primer ciclo va a recorrer cada fila de mi matriz.
- El segundo ciclo va a recorrer los números que se encuentran dentro de cada fila.

Voy a utilizar `print(numero, end="\t")` para que cada número aparezca separado por una tabulación. Esto permite que los resultados se vean organizados en columnas y no uno debajo del otro.

También voy a utilizar un `print()` vacío al terminar cada fila. Esto hace que la siguiente fila aparezca en una nueva línea.

En esta función no voy a utilizar `return`, porque no necesito regresar ningún resultado. Solamente quiero mostrar la información en pantalla.

```python

# PARTE B: FUNCIÓN PARA MOSTRAR LA MATRIZ

# Defino una función que recibe mi matriz como parámetro.
def imprimir_tabla(tabla):

    # Recorro cada fila que existe dentro de la matriz.
    for fila in tabla:

        # Recorro cada número que se encuentra
        # dentro de la fila actual.
        for numero in fila:

            # Muestro cada número y utilizo una tabulación
            # para separar los valores y formar las columnas.
            print(numero, end="\t")

        # Al terminar una fila, hago un salto de línea
        # para que la siguiente aparezca debajo.
        print()


# Muestro un título antes de imprimir la tabla.
print("TABLA DE PITÁGORAS DEL 1 AL 10")
print("-------------------------------------------")

# Llamo a mi función y le envío la matriz que construí.
imprimir_tabla(tabla_Pitagoras)
```

## PARTE C. FUNCIÓN PARA CONSULTAR UN PRODUCTO

En esta parte voy a crear una segunda función llamada `consultar_producto(tabla, renglon, columna)`.

A diferencia de la función anterior, esta sí va a utilizar `return`, porque necesito que me entregue el resultado de la posición que estoy consultando.

La función recibirá tres datos:

- `tabla`: es la matriz donde guardé todos los resultados.
- `renglon`: es el número de la fila que quiero consultar.
- `columna`: es el número de la columna que quiero consultar.

Para encontrar el resultado voy a utilizar los índices de la matriz.

En Python las posiciones de las listas comienzan desde cero, mientras que mi tabla comienza desde el número uno. Por eso voy a restar uno al renglón y también a la columna.

Por ejemplo, si quiero consultar el renglón 7 y la columna 8, Python debe buscar la posición `[6][7]`.

Es importante mencionar que en esta función no voy a utilizar el operador `*`. No necesito volver a multiplicar los números, porque el resultado ya se encuentra guardado dentro de mi matriz.

Finalmente, utilizaré `return` para regresar el valor que encontré.

```python

# PARTE C: FUNCIÓN PARA CONSULTAR UN PRODUCTO

# Defino una función que recibe la matriz,
# el renglón y la columna que quiero consultar.
def consultar_producto(tabla, renglon, columna):

    # Busco directamente el resultado dentro de la matriz.
    # Resto uno al renglón y a la columna porque
    # las posiciones de las listas en Python comienzan en cero.
    resultado = tabla[renglon - 1][columna - 1]

    # Regreso el resultado que encontré en la matriz.
    return resultado
```

## PARTE D. PROGRAMA PRINCIPAL Y VALIDACIÓN DE DATOS

Ahora voy a unir las funciones que desarrollé para que mi programa pueda recibir datos del usuario y mostrar el resultado de la consulta.

Primero voy a pedir que se ingrese el número del renglón y después el número de la columna. Para capturar estos datos voy a utilizar `input()` y convertiré las respuestas a números enteros con `int()`.

Después voy a utilizar dos ciclos `while` para validar los datos.

La primera validación revisará que el renglón esté entre 1 y 10. La segunda hará lo mismo con la columna.

Si el usuario escribe un número menor que 1 o mayor que 10, el programa mostrará un mensaje de advertencia y volverá a pedir el dato hasta que sea correcto.

Esta validación es importante porque mi matriz solamente tiene 10 filas y 10 columnas. Si permitiera consultar una posición fuera del rango, podría obtener un resultado incorrecto.

Cuando los dos números sean válidos, voy a llamar a mi función `consultar_producto()` y guardaré el resultado que me regrese en una variable llamada `resultado_final`.

Por último, mostraré el resultado utilizando un mensaje que indique los dos números consultados y el producto encontrado.

```python

# PARTE D: PROGRAMA PRINCIPAL

# Solicito al usuario el número del renglón que desea consultar.
# Utilizo int() para convertir la respuesta en un número entero.
renglon = int(input("Ingresa el número del renglón (del 1 al 10): "))

# Compruebo que el renglón esté dentro del rango permitido.
# Mientras el número sea menor que 1 o mayor que 10,
# el programa seguirá solicitando un nuevo valor.
while renglon < 1 or renglon > 10:

    # Muestro un aviso cuando el número no es válido.
    print("El número del renglón está fuera del rango permitido.")

    # Vuelvo a solicitar el renglón hasta recibir uno correcto.
    renglon = int(input("Intenta nuevamente. Ingresa un número del 1 al 10: "))


# Solicito al usuario el número de la columna que desea consultar.
columna = int(input("Ingresa el número de la columna (del 1 al 10): "))

# Compruebo que la columna también esté dentro del rango.
while columna < 1 or columna > 10:

    # Muestro un aviso si la columna no es válida.
    print("El número de la columna está fuera del rango permitido.")

    # Solicito nuevamente la columna hasta que sea correcta.
    columna = int(input("Intenta nuevamente. Ingresa un número del 1 al 10: "))


# Llamo a mi función de consulta y le envío la matriz,
# el renglón y la columna que ingresó el usuario.
# Guardo el resultado que regresa la función.
resultado_final = consultar_producto(tabla_Pitagoras, renglon, columna)


# Muestro el resultado final de la consulta.
print("\nRESULTADO DE LA CONSULTA")

# Presento los dos números consultados y el resultado encontrado.
print(f"{renglon} x {columna} = {resultado_final}")
```

## PARTE E. REALIZACIÓN DE LOS CASOS DE PRUEBA

En esta última parte voy a comprobar que mi programa funcione correctamente mediante cuatro casos de prueba.

Mi intención es revisar diferentes situaciones para asegurarme de que la matriz guarde los resultados correctos y que mi función pueda consultarlos sin realizar nuevamente las multiplicaciones.

Voy a comprobar los siguientes casos:

**Caso 1. Consulta de 7 × 8**

Voy a consultar el renglón 7 y la columna 8. Espero obtener como resultado 56.

**Caso 2. Consulta de 1 × 1**

Voy a consultar el primer renglón y la primera columna. Espero obtener como resultado 1.

**Caso 3. Consulta de 10 × 10**

Voy a consultar el último renglón y la última columna de mi matriz. Espero obtener como resultado 100.

**Caso 4. Validación de un número fuera del rango**

Voy a simular que el usuario ingresa primero el número 0 para el renglón. Como este número no está permitido, espero que el programa muestre un mensaje de advertencia. Después voy a ingresar el número 7 y utilizaré la columna 8. Espero que el programa acepte los datos correctos y muestre como resultado 56.

Para evitar que los resultados de una prueba se borren cuando realice la siguiente, voy a ejecutar los cuatro casos desde una sola celda. De esta manera podré observar todos los resultados juntos.

También voy a comparar el resultado esperado con el resultado obtenido para comprobar si cada prueba fue correcta.

```python

# PARTE E: COMPROBACIÓN DE LOS CUATRO CASOS DE PRUEBA

# Primero creo una lista con los tres primeros casos.
# Cada caso contiene el renglón, la columna y el resultado esperado.
casos_prueba = [
    (7, 8, 56),
    (1, 1, 1),
    (10, 10, 100)
]

# Recorro los tres casos de prueba que guardé en mi lista.
for numero_caso, caso in enumerate(casos_prueba, start=1):

    # Guardo el número del renglón que voy a consultar.
    renglon_prueba = caso[0]

    # Guardo el número de la columna que voy a consultar.
    columna_prueba = caso[1]

    # Guardo el resultado que espero obtener.
    resultado_esperado = caso[2]

    # Consulto el resultado utilizando mi función.
    # De esta manera no vuelvo a realizar la multiplicación.
    resultado_obtenido = consultar_producto(
        tabla_Pitagoras,
        renglon_prueba,
        columna_prueba
    )

    # Comparo el resultado obtenido con el resultado esperado.
    if resultado_obtenido == resultado_esperado:

        # Si los dos resultados son iguales,
        # significa que la prueba pasó correctamente.
        estado = "CORRECTO"

    else:

        # Si los resultados son diferentes,
        # significa que debo revisar mi programa.
        estado = "INCORRECTO"

    # Muestro la información completa de cada prueba.
    print(f"CASO DE PRUEBA {numero_caso}")
    print("Renglón:", renglon_prueba)
    print("Columna:", columna_prueba)
    print("Resultado esperado:", resultado_esperado)
    print("Resultado obtenido:", resultado_obtenido)
    print("Estado:", estado)
    print("----------------------------------------")


# CUARTO CASO: VALIDACIÓN DE UN NÚMERO FUERA DEL RANGO

print("CASO DE PRUEBA 4")

# Simulo que el usuario ingresa primero el número 0.
renglon_prueba = 0

# Guardo el valor correcto que se ingresará después del aviso.
nuevo_renglon = 7

# Compruebo el dato utilizando la misma condición
# de validación que utilicé en mi programa principal.
while renglon_prueba < 1 or renglon_prueba > 10:

    # Muestro el mensaje porque el número 0 no es válido.
    print("El número del renglón está fuera del rango permitido.")

    # Vuelvo a asignar el renglón con un valor válido.
    renglon_prueba = nuevo_renglon

# Muestro el renglón que finalmente fue aceptado.
print("Renglón válido:", renglon_prueba)

# Utilizo la columna 8 para completar esta consulta.
columna_prueba = 8

# Defino el resultado que espero obtener.
resultado_esperado = 56

# Consulto el resultado usando mi función.
resultado_obtenido = consultar_producto(
    tabla_Pitagoras,
    renglon_prueba,
    columna_prueba
)

# Comparo el resultado obtenido con el esperado.
if resultado_obtenido == resultado_esperado:
    estado = "CORRECTO"
else:
    estado = "INCORRECTO"

# Muestro los resultados de la cuarta prueba.
print("Columna:", columna_prueba)
print("Resultado esperado:", resultado_esperado)
print("Resultado obtenido:", resultado_obtenido)
print("Estado:", estado)
print("----------------------------------------")
```

En el cuarto caso comprobé la validación de los datos. Primero utilicé el número 0, que se encuentra fuera del rango permitido. El programa mostró el mensaje de advertencia y después aceptó el número 7 como valor correcto. Al consultar la columna 8, obtuve nuevamente el resultado 56.
